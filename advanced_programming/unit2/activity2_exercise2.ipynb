```python
import sys
import csv
import ImportCSV

records = ImportCSV.get_records("PeopleTrainingDate.csv")

if len(records) == 0:
    print("No data")
    sys.exit()

# Sort by column Updated.
# The date is reversed from d/m/Y to Ymd format for proper sorting.
records.sort(key=lambda record: "".join(reversed(record["Updated"].split("/"))))

# Open an output file for writing, the file will be overwritten.
# No newline, as csv writer outputs its own newline.
with open("PeopleTraining_Output.csv", "w", newline="") as file:
    csv_writer = csv.writer(file)

    cols = ["Updated", "Title", "Name", "ID", "Email", "Company"]

    csv_writer.writerow(cols)

    for record in records:
        row = [record[col] for col in cols]
        csv_writer.writerow(row)

print(str(len(records)) + " rows output")
```
