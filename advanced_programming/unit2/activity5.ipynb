```python
import json
import xml.etree.ElementTree as ET
from collections import defaultdict

xml_tree = ET.parse("People.xml")

xml_root = xml_tree.getroot()

json_elements = {xml_root.tag: []}

for xml_elem in xml_root:
    full_name = xml_elem.find("fullName")
    json_name = {}

    if full_name is not None:
        title = full_name.get("title")
        if title is not None:
            json_name["title"] = title

        first_name = full_name.findtext("firstName")
        if first_name is not None:
            json_name["firstName"] = first_name

        surname = full_name.findtext("surname")
        if surname is not None:
            json_name["surname"] = surname
        
        other = full_name.find("other")
        if other is not None:
            json_name["other"] = []
            for other_name in other.findall("name"):
                json_name["other"].append(other_name.text)

    json_elements[xml_root.tag].append({
        "fullName": json_name,
        "age": xml_elem.findtext("age"),
        "city": xml_elem.findtext("city")
    })

json.dump(json_elements, open("People_converted.json", "w"), indent=4)

with open("People_converted.json") as file:
    print(file.read())
```
