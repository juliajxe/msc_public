```python
import re

with open("The_Raven.txt", encoding="utf8", newline="\n") as file:
    text = file.read()

    # Match the word shrieked between two word boundaries.
    print(re.search(r"\bshrieked\b", text))

    # Match the word bleak between two word boundaries.
    print(re.search(r"\bbleak\b", text))

    # Match, between word boundaries, any number of alphanum with pp
    # anywhere in the middle.
    words_with_pp = re.findall(r"\b\w*pp\w*\b", text)
    print(len(words_with_pp), words_with_pp)

    # Replace ! with #
    re.sub("!", "#", text)

    # Match, between word boundaries, the letter t followed by
    # any number of alphanum followed by any character that is neither
    # an e nor a non-alphanum (i.e. a valid word character but not e)
    print(re.findall(r"\bt\w*[^e\W]\b", text, re.IGNORECASE))
```
