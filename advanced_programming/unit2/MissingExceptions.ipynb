to find the type of error/s that are generated

```python
try:
  num1 = int(input("Enter a number:" ))
  num2 = int(input("Enter another number:" ))
except ValueError:
  print("Only integer numbers are accepted")
except OverflowError:
  print("Number too large")

try:
  if num1 > num2:
    if num1%num2 == 0:
      print(num1, " is a multiple of ", num2)
  else:
    if num2%num1 == 0:
      print(num2, " is a multiple of ", num1)
except ZeroDivisionError:
  print("Zero is not allowed")
```

validating user given file names, what can go wrong?

```python
fileName = input("Enter File:")

try:
    print(open(fileName).read())
except FileNotFoundError:
    print("File not found")
except PermissionError:
    print("You do not have access to that file")
except IsADirectoryError:
    print("A directory not a file was provided")
except OSError as err:
    print(err.strerror)
```
